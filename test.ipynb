```python
couverture = [25, 45, 6, 8, 12]
support_couverture = [12, 10, 4, 12, 0]
couche_1 = [2.8, 0]
couche_2 = [3.45, 5, 0]
Finition = [17]
marge = 5

value = []

for i1 in couverture:
    for i2 in support_couverture:
        for i3 in couche_1:
            for i4 in couche_2:
                for i5 in Finition:
                    value.append(round(i1 + i2 + i3 + i4 + i5 + marge))

value = sorted(list(set(value)))
print(value)
```
